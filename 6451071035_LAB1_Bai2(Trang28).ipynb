Câu 2. Lấy ra 5 thuộc tính có số lượng giá trị thiếu nhiều nhất

```python

print("Phạm Lê Anh Khoa - 6451071035")

df = df.replace('unknown', np.nan)

total = df.isnull().sum().sort_values(ascending=False)
percent_1 = df.isnull().sum() / len(df) * 100
percent_2 = round(percent_1, 1).sort_values(ascending=False)

missing_data = pd.concat(
    [total, percent_2],
    axis=1,
    keys=['Total', '%']
)

missing_data.head(5)
```

Câu 5. Tiền xử lý dữ liệu bị thiếu

1. Xử lý các thuộc tính định tính

```python

print("Phạm Lê Anh Khoa - 6451071035")

categorical_cols = df.select_dtypes(include=['object']).columns

for col in categorical_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].mode()[0])

df.isnull().sum()
```

2. Xử lý các thuộc tính định lượng

```python
print("Phạm Lê Anh Khoa - 6451071035")
numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    if df[col].isnull().sum() > 0:
        df[col] = df[col].fillna(df[col].median())

df.isnull().sum()
```

Câu 6. Mã hóa thuộc tính định tính và chuẩn hóa dữ liệu

1. Chuyển đổi thuộc tính quyết định y thành số

```python

print("Phạm Lê Anh Khoa - 6451071035")

df['y'] = df['y'].map({
    'no': 0,
    'yes': 1
})

df['y'].value_counts()
```

2. Mã hóa các thuộc tính định tính

```python
print("Phạm Lê Anh Khoa - 6451071035")
df_encoded = pd.get_dummies(
    df,
    columns=df.select_dtypes(include=['object']).columns,
    dtype=int
)

df_encoded.head()
```

3. Chuẩn hóa dữ liệu bằng Min-Max Normalization

```python
print("Phạm Lê Anh Khoa - 6451071035")
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

numeric_cols = df_encoded.select_dtypes(
    include=np.number
).columns.drop('y')

df_encoded[numeric_cols] = scaler.fit_transform(
    df_encoded[numeric_cols]
)

df_encoded.head(12)
```
