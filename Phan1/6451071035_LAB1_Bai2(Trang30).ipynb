Câu a) Thêm vào tiêu đề các cột theo thứ tự: id, device_id, co_level, humidity, temperature, time

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Phạm Lê Anh Khoa - 6451071035")

with open('data/tb_tracking.xlsx', 'rb') as f:
    df = pd.read_excel(f, header=None)

df.columns = [
    'id',
    'device_id',
    'co_level',
    'humidity',
    'temperature',
    'time'
]

df.head(12)
```

Câu b) Loại bỏ những dữ liệu đo trước ngày 03/01/2018

```python

print("Phạm Lê Anh Khoa - 6451071035")

df['time'] = pd.to_datetime(
    df['time'],
    dayfirst=True,
    errors='coerce'
)

df = df[
    df['time'] >= pd.Timestamp('2018-01-03')
].copy()

df.head(12)
```

Câu c) Từ cột time bóc tách thời điểm đo đạc trong ngày tính theo phút, đặt tên là minutes

```python

  print("Phạm Lê Anh Khoa - 6451071035")

  df['minutes'] = (
      df['time'].dt.hour * 60
      + df['time'].dt.minute
  )

  df[['time', 'minutes']].head(12)
```

Câu d) Vẽ đồ thị phân tán (scatter-plots) biểu diễn nồng độ khí CO, độ ẩm và nhiệt độ theo cột minutes

```python

print("Phạm Lê Anh Khoa - 6451071035")

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.scatterplot(
    data=df,
    x='minutes',
    y='co_level',
    ax=axes[0]
)
axes[0].set_title('CO Level vs Minutes')

sns.scatterplot(
    data=df,
    x='minutes',
    y='humidity',
    ax=axes[1]
)
axes[1].set_title('Humidity vs Minutes')

sns.scatterplot(
    data=df,
    x='minutes',
    y='temperature',
    ax=axes[2]
)
axes[2].set_title('Temperature vs Minutes')

plt.tight_layout()
plt.show()
```
