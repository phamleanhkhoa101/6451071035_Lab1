Câu 6. Tiền xử lý dữ liệu thiếu và các thuộc tính không cần thiết

    Kiểm tra các thuộc tính thiếu dữ liệu

```python
print("Phạm Lê Anh Khoa - 6451071035")

df.isin(['unknown']).sum()
```

    Chuyển các giá trị unknown thành NaN

```python
print("Phạm Lê Anh Khoa - 6451071035")
df_clean = df.copy()

df_clean = df_clean.replace('unknown', np.nan)

df_clean.isnull().sum()
```

    Xử lý các thuộc tính định tính bị thiếu

```python
print("Phạm Lê Anh Khoa - 6451071035")
categorical_cols = df_clean.select_dtypes(
    include=['object', 'str']
).columns

for col in categorical_cols:
    if df_clean[col].isnull().sum() > 0:
        df_clean[col] = df_clean[col].fillna(
            df_clean[col].mode()[0]
        )

df_clean.isnull().sum()
```

    Xử lý các thuộc tính định lượng bị thiếu

```python
print("Phạm Lê Anh Khoa - 6451071035")
numeric_cols = df_clean.select_dtypes(
    include=np.number
).columns

for col in numeric_cols:
    if df_clean[col].isnull().sum() > 0:
        df_clean[col] = df_clean[col].fillna(
            df_clean[col].median()
        )

df_clean.isnull().sum()
```

Kiểm tra lại dữ liệu sau khi xử lý

```python
print("Phạm Lê Anh Khoa - 6451071035")
df_clean.info()
```

Câu 7. Biến đổi các thuộc tính dữ liệu

    Chuyển đổi thuộc tính quyết định y thành số

```python
print("Phạm Lê Anh Khoa - 6451071035")

df_clean['y'] = df_clean['y'].map({
    'no': 0,
    'yes': 1
})

df_clean['y'].value_counts()
```

Chuyển đổi thuộc tính default, housing, loan thành số

```python
print("Phạm Lê Anh Khoa - 6451071035")
binary_mapping = {
    'no': 0,
    'yes': 1
}

for col in ['default', 'housing', 'loan']:
    df_clean[col] = df_clean[col].map(binary_mapping)

df_clean[['default', 'housing', 'loan']].head(10)
```

Chia thuộc tính tuổi (age) thành 8 nhóm

```python
print("Phạm Lê Anh Khoa - 6451071035")

df_clean['age_group'] = pd.qcut(
    df_clean['age'],
    q=8,
    labels=False,
    duplicates='drop'
)

df_clean['age_group'].value_counts().sort_index()
```

Chia thuộc tính số dư (balance) thành 6 nhóm

```python
print("Phạm Lê Anh Khoa - 6451071035")
df_clean['balance_group'] = pd.qcut(
    df_clean['balance'],
    q=6,
    labels=False,
    duplicates='drop'
)

df_clean['balance_group'].value_counts().sort_index()
```

Chuyển các thuộc tính định tính về One-hot Encoding

```python
print("Phạm Lê Anh Khoa - 6451071035")
df_encoded = pd.get_dummies(
    df_clean,
    columns=df_clean.select_dtypes(
        include=['object', 'str']
    ).columns,
    dtype=int
)

df_encoded.head(10)
```

Chuẩn hóa các thuộc tính định lượng bằng Min-Max Normalization

```python
print("Phạm Lê Anh Khoa - 6451071035")

scaler = MinMaxScaler()

numeric_cols = [
    'age', 'balance', 'day',
    'duration', 'campaign', 'pdays', 'previous'
]

df_encoded[numeric_cols] = scaler.fit_transform(
    df_encoded[numeric_cols]
)

df_encoded.head(10)
```

Câu 8. Tạo các thuộc tính mới

    Tạo thuộc tính tổng số lần liên hệ (total_contacts)

```python
print("Phạm Lê Anh Khoa - 6451071035")

df_clean['total_contacts'] = (
    df_clean['campaign'] + df_clean['previous']
)

df_clean[
    ['campaign', 'previous', 'total_contacts']
].head(10)
```

    Tạo thuộc tính thời lượng trung bình trên mỗi lần liên hệ (duration_per_contact)

```python
print("Phạm Lê Anh Khoa - 6451071035")
df_clean['duration_per_contact'] = (
    df_clean['duration'] / df_clean['campaign'].clip(lower=1)
)

df_clean[
    ['duration', 'campaign', 'duration_per_contact']
].head(10)
```

    Tạo thuộc tính khách hàng có khoản vay hay không (has_loan)

```python
print("Phạm Lê Anh Khoa - 6451071035")
df_clean['has_loan'] = (
    (df_clean['housing'] == 1) |
    (df_clean['loan'] == 1)
).astype(int)

df_clean[
    ['housing', 'loan', 'has_loan']
].head(10)
```

    Tạo thuộc tính số dư trung bình theo tuổi (balance_per_age)

```python
print("Phạm Lê Anh Khoa - 6451071035")
df_clean['balance_per_age'] = (
    df_clean['balance'] / df_clean['age'].clip(lower=1)
)

df_clean[
    ['age', 'balance', 'balance_per_age']
].head(10)
```
